```python
# jointplot
sns.jointplot(x='total_bill',y= 'tip',data=tips,kind='hex',height=8)
plt.xlabel('Total Bill',fontsize=14)
plt.ylabel('Tips',fontsize=14)
plt.show()
```

```python
# jointplot
sns.jointplot(x='total_bill',y= 'tip',data=tips,kind='reg',height=8)
plt.xlabel('Total Bill',fontsize=14)
plt.ylabel('Tips',fontsize=14)
plt.show()
```

## Seaborn Matrix Plots

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
tips.head()
```

```python
flights.head()
```

```python
tips.corr(numeric_only= True)
```

```python
tc=tips.corr(numeric_only= True)
```

```python
fig=plt.figure(figsize=(12,6))
sns.heatmap(data=tc)
plt.show()
```

```python
fig=plt.figure(figsize=(12,6))
sns.heatmap(data=tc,annot= True)
plt.show()
```

```python
fig=plt.figure(figsize=(12,6))
sns.heatmap(data=tc,annot= True,cmap='coolwarm')
plt.show()
```

```python
flights.pivot_table(index ='month', columns='year',values='passengers',observed= False)
```

```python
fp
```

## Seaborn grids

```python
iris =sns.load_dataset('iris')
```

```python
iris.head()
```

```python
iris['species'].unique()
```

```python
sns.pairplot(iris)
plt.show()
```

```python
g=sns.PairGrid(iris)
g.map(plt.scatter)
plt.show()
```

```python
g=sns.PairGrid(iris)
g.map_diag(sns.histplot,kde =True)
plt.show()
```

```python
g=sns.PairGrid(iris)
g.map_diag(sns.histplot,kde =True)
g.map_upper(plt.scatter)
plt.show()
```

```python
g=sns.PairGrid(iris)
g.map_diag(sns.histplot,kde =True)
g.map_upper(plt.scatter)
g.map_lower(sns.kdeplot,color='b',fill=True)
plt.show()
```
