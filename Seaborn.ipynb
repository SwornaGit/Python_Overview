```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
tips= sns.load_dataset('tips')
```

```python
tips
```

```python
tips.head()
```

```python
sns.displot(data=tips, x='total_bill')
plt.show()
```

```python
sns.displot(data=tips, x='total_bill',kde=True)
plt.show()
```

```python
sns.displot(data=tips, x='total_bill',kde=True, bins=50)
plt.show()
```

```python
# jointplot
sns.jointplot(x='total_bill',y= 'tip',data=tips,height=8)
plt.show()
```

```python
# jointplot
sns.jointplot(x='total_bill',y= 'tip',data=tips,kind='hex',height=8)
plt.show()
```

```python
# jointplot
sns.jointplot(x='total_bill',y= 'tip',data=tips,kind='hex',height=8)
plt.xlabel('Total Bill',fontsize=14)
plt.ylabel('Tips',fontsize=14)
plt.show()
```

```python
# jointplot
sns.jointplot(x='total_bill',y= 'tip',data=tips,kind='reg',height=8)
plt.xlabel('Total Bill',fontsize=14)
plt.ylabel('Tips',fontsize=14)
plt.show()
```

```python
# jointplot
sns.jointplot(x='total_bill',y= 'tip',data=tips,kind='kde',height=8)
plt.xlabel('Total Bill',fontsize=14)
plt.ylabel('Tips',fontsize=14)
plt.show()
```

```python
# jointplot
sns.jointplot(x='total_bill',y= 'tip',data=tips,kind='kde',height=8,color='skyblue', fill=True)
plt.xlabel('Total Bill',fontsize=14)
plt.ylabel('Tips',fontsize=14)
plt.show()
```

```python
sns.pairplot(data=tips)
plt.show()
```

```python
sns.pairplot(data=tips,hue='sex')
plt.show()
```

```python
sns.pairplot(data=tips,hue='sex',palette='coolwarm')
plt.show()
```

```python
sns.pairplot(data=tips,hue='sex',kind='hist')
plt.show()
```

```python
sns.rugplot(tips['total_bill'])
plt.show()
```

```python
sns.histplot(tips['total_bill'])
plt.show()
```

```python
sns.barplot(tips['total_bill'])
plt.show()
```

```python
sns.kdeplot(tips['total_bill'])
plt.show()
```

## Seaborn categorical Plots

```python
tips.head()
```

```python
sns.barplot(data = tips,x ='sex',y='total_bill')
plt.show()
```

```python
sns.countplot(tips['sex'])
plt.show()
```

```python
sns.countplot(x=tips['sex'])
plt.show()
```

```python
sns.boxplot(x='day',y='total_bill',data=tips)
plt.show()
```

```python
tips.head(10)
```

```python
sns.boxplot(x='day',y='total_bill',data=tips, hue='smoker')
plt.show()
```

```python
sns.violinplot(x='day',y='total_bill',data = tips)
plt.show()
```

```python
fig = plt.figure(figsize = (12,6))
sns.violinplot(x='day',y='total_bill',data = tips)
plt.show()
```

```python
fig = plt.figure(figsize = (12,6))
sns.violinplot(x='day',y='total_bill',data = tips,hue='sex',width=0.8)
plt.show()
```

```python
fig = plt.figure(figsize = (12,6))
sns.violinplot(x='day',y='total_bill',data = tips,hue='sex',split=True)
plt.show()
```

```python
fig = plt.figure(figsize = (8,6))
sns.stripplot(x='day',y='total_bill',data = tips)
plt.show()
```

```python
fig = plt.figure(figsize = (8,6))
sns.stripplot(x='day',y='total_bill',data = tips, hue='sex',dodge=True) #dodge instade of split
plt.tight_layout()
plt.show()
```

```python
fig = plt.figure(figsize = (8,6))
sns.stripplot(x='day',y='total_bill',data = tips, hue='sex') 
plt.tight_layout()
plt.show()
```

```python
sns.swarmplot(x='day',y= 'total_bill',data=tips,hue='sex')
plt.tight_layout()
plt.show()
```

```python
sns.swarmplot(x='day',y= 'total_bill',data=tips,hue='sex',dodge=True)
plt.tight_layout()
plt.show()
```

```python
fig =plt.figure(figsize=(12,6))
sns.violinplot(x=tips['day'],y=tips['total_bill'])
sns.swarmplot(x=tips['day'],y=tips['total_bill'],color='black')
plt.show()
```
